## 2. 전처리-변경

```python
import pandas as pd
import re
import random

def safe_preprocess(text):
    
    # 문맥 추론에 필요한 최소한의 구두점은 보존하는 정규표현식
    text = re.sub(r"[^가-힣0-9a-zA-Z\sㅋㅎ~]", "", str(text))
    text = re.sub(r"(ㅋ|ㅎ){3,}", r"\1\1", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

# 데이터 로드
train = pd.read_csv('train.csv')

# 입력과 출력 데이터 모두 동일한 정제 기준을 적용하여 토큰 공간(Token Space)을 일치시킵니다.
train['input'] = train['input'].apply(safe_preprocess)
train['output'] = train['output'].apply(safe_preprocess)
```

## 3. 데이터 증강

```python
import random

def final_augment(text):
    words = text.split()
    new_words = []

    for i, w in enumerate(words):
        
        # 👉 일부 단어만 변형 (핵심)
        if random.random() < 0.2:
            
            # 1. 약한 띄어쓰기 (확률 낮춤)
            if random.random() < 0.15:
                w = " ".join(list(w))
            
            # 2. 초성 느낌 (간단 버전)
            if random.random() < 0.1 and len(w) > 1:
                w = w[0]
            
            # 발음 변형 강화
            if random.random() < 0.2:
                w = w.replace("요", "용").replace("다", "타").replace("네", "넹")

            # 오타 추가
            if random.random() < 0.15:
                w = w.replace("좋", "조").replace("가", "갸")

            # 공백 조금 더
            if random.random() < 0.15:
                w = " ".join(list(w))

            if random.random() < 0.2:
                w = w.replace("ㄱ", "ㅋ").replace("ㄷ", "ㅌ").replace("ㅂ", "ㅍ")

            if random.random() < 0.2:
                w = w.replace("ㄱ", "ㄲ").replace("ㄷ", "ㄸ").replace("ㅂ", "ㅃ") 

        # 👉 특수문자는 문장 끝에만
        if i == len(words) - 1 and random.random() < 0.3:
            w = w + random.choice(["!", "~"])
        
        new_words.append(w)

    return " ".join(new_words)
```

## 4. LLM용 데이터셋

```python
# 1. 증강용 데이터프레임 복제 (원본 보존)
aug_df = train.copy()

# 2. 정답(output)을 기반으로 인위적 노이즈(input) 생성
aug_df['input'] = aug_df['output'].apply(final_augment)

# 3. 데이터 결합 (원본 데이터 + 증강 데이터)
# 이 과정을 통해 데이터셋의 볼륨이 2배로 증가하며, 원본 분포를 잃지 않습니다.
final_train = pd.concat([train, aug_df], ignore_index=True)

# 4. 데이터 셔플링 (Shuffling)
# 모델이 학습 순서에 과적합(Overfitting)되지 않도록 데이터를 무작위로 섞어줍니다.
final_train = final_train.sample(frac=1, random_state=42).reset_index(drop=True)

print(f"최종 병합된 데이터 개수: {len(final_train)}")
```

```python
# for문 없이 한 줄로 List of Dictionaries 형태로 변환
dataset = final_train[['input', 'output']].to_dict('records')

# 샘플 확인
print("샘플 데이터 확인:")
for item in dataset[:3]:
    print(item)

# CSV로 최종 저장
final_train[['input', 'output']].to_csv("augmented_train_v2.csv", index=False)
```
