# Team A 데이터 분석 / 전처리 / 증강 코드

## 1. 데이터 확인

```python
import pandas as pd
train = pd.read_csv('train.csv')

for i in range(10):
    print('입력:', train['input'][i])
    print('정답:', train['output'][i])
    print('-'*50)
```

## 2. 전처리

```python
import re

def preprocess(text):
    text = re.sub(r"[^가-힣0-9a-zA-Z\sㅋㅎ]", "", text)
    text = re.sub(r"(ㅋ|ㅎ){3,}", r"\1\1", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

train['input'] = train['input'].apply(preprocess)
print(train['input'].head())
```

## 3. 데이터 증강

```python
import random

def final_augment(text):
    words = text.split()
    new_words = []

    for i, w in enumerate(words):
        
        # 👉 일부 단어만 변형 (핵심)
        if random.random() < 0.2:
            
            # 1. 약한 띄어쓰기 (확률 낮춤)
            if random.random() < 0.15:
                w = " ".join(list(w))
            
            # 2. 초성 느낌 (간단 버전)
            if random.random() < 0.1 and len(w) > 1:
                w = w[0]
            
            # 발음 변형 강화
            if random.random() < 0.2:
                w = w.replace("요", "용").replace("다", "타").replace("네", "넹")

            # 오타 추가
            if random.random() < 0.15:
                w = w.replace("좋", "조").replace("가", "갸")

            # 공백 조금 더
            if random.random() < 0.15:
                w = " ".join(list(w))

            if random.random() < 0.2:
                w = w.replace("ㄱ", "ㅋ").replace("ㄷ", "ㅌ").replace("ㅂ", "ㅍ")

            if random.random() < 0.2:
                w = w.replace("ㄱ", "ㄲ").replace("ㄷ", "ㄸ").replace("ㅂ", "ㅃ") 

        # 👉 특수문자는 문장 끝에만
        if i == len(words) - 1 and random.random() < 0.3:
            w = w + random.choice(["!", "~"])
        
        new_words.append(w)

    return " ".join(new_words)
```

## 4. LLM용 데이터셋

```python
# 1. 데이터 로드
#train = pd.read_csv(train.csv)

# 2. 전처리
train['input'] = train['input'].apply(preprocess)

# ✅ 3. augmentation (여기 추가!!!)
train['aug_input'] = train['output'].apply(final_augment)
print(train[['input', 'aug_input']].head())
```

```python
dataset = []

for i in range(len(train)):
    dataset.append({
        'input': train['aug_input'][i],
        'output': train['output'][i]
    })

dataset[:3]
```
